# IAM: Intelligent Assistant for Molecules

IAM is an AI-powered chemistry assistant platform designed for energetic materials research, offering automated workflows for molecular modeling, property prediction, sensitivity estimation, and data extraction. IAM is modular, Raspberry Pi compatible, and built to evolve with the user's research needs.

## System Architecture Overview

```
IAM/
├── IAM_GUI/               ← Web interface (Flask + HTML/CSS/JS + 3Dmol.js)
├── IAM_Engine/            ← Core chemistry backend (RDKit + XTB + Psi4)
├── IAM_Knowledge/         ← Molecule database, results, JSON files
├── IAM_Predictors/        ← ML + Empirical prediction models
├── IAM_Agent.py           ← Automation watcher for .xyz & dataset updates
├── chem-env.yaml          ← Conda environment
├── start_all.sh           ← Launches all services
└── IAM_Master_Energetics.xlsx ← Parsed master dataset (423+ compounds)
```

## Modules

### Property & Performance Prediction
- Kamlet-Jacobs
- Keshavarz empirical formulas
- ML model trained on Klapötke, Agrawal, LLNL

### Sensitivity Prediction
- Impact, friction, heat, electrostatic shock
- SMARTS-based substructure detection
- Hydrolysis scoring

### Web GUI (IAM Molecule Viewer)
- Draw/import: SMILES / MOL / XYZ
- Visualize in 3D with 3Dmol.js
- Embedded Ketcher editor
- Tabs for Results / Logs / Settings
- Options for charge, multiplicity, GFN model
- Dynamic tooltips and output fields

### Dataset Extractor
- Automated parsing from:
  - Klapötke Encyclopedias Vol. A–D, E, O
  - Agrawal's High Energy Materials
  - Keshavarz Performance Prediction
- Output: structured .csv/.json
- 423+ entries parsed with full properties

### Automation
- `IAM_Agent.py`: watches `ToAnalyze/` folder
- Automatically runs XTB, parses results
- Updates master benchmark
- Creates output zip with JSON, .log, .cube

## Installation Steps

**Requirements**: Conda, Linux/WSL/Raspberry Pi, ~4 GB RAM minimum

```bash
# Clone the repo and run:
bash install_iam_node.sh
source activate chem-env
bash start_all.sh
```

Access the interface at: [http://localhost:5000](http://localhost:5000)
